```python
### Read in snapshot polygon and fireline
snap = fv.prep_gdf(date = '20201031PM',layer='perimeter',handle_multi=True,only_lf=False,area_lim=4.7, year = "2020", path_region = "WesternUS")
gdf_test = fv.load_file(date = '20201031PM',layer="fireline",handle_multi=False,only_lf=False,area_lim=4.7, year = "2020", path_region = "WesternUS")

gdf_test["t"] = gdf_test["t"].astype("str") ## Make fireline .explore compatable

snap = snap[snap.t >= "2020-10-26 12:00:00"] # Only recent fires
gdf_test = gdf_test[gdf_test.t >= "2020-10-26 12:00:00"] 
gdf_test = gdf_test[gdf_test.fireID.isin(snap.index)] # Only the large fires 

m = snap.explore()
m = gdf_test.explore(m = m, color = "red") # active fireline in red, polygon in blue
m
```
